# SpaceX API data collection and wrangling

The live SpaceX API returned HTTP 525 during this run. This reproducible version loads the fixed IBM course snapshot and records the source explicitly.

```python
from pathlib import Path
import pandas as pd
import requests
```

```python
spacex_url = "https://api.spacexdata.com/v4/launches/past"
# When the live endpoint is available: requests.get(spacex_url, timeout=20).json()
# The supplied snapshot preserves the course’s historical date range.
data = pd.read_csv(Path("dataset_part_1.csv"))
data.shape
```

```python
data[["FlightNumber", "Date", "BoosterVersion", "PayloadMass", "Orbit", "LaunchSite", "Outcome"]].head()
```

```python
data_falcon9 = data.loc[data["BoosterVersion"].eq("Falcon 9")].copy()
data_falcon9["FlightNumber"] = range(1, len(data_falcon9) + 1)
data_falcon9.shape
```

```python
data_falcon9.isnull().sum()
```

```python
data_falcon9["PayloadMass"] = data_falcon9["PayloadMass"].fillna(data_falcon9["PayloadMass"].mean())
data_falcon9.to_csv("dataset_part_1_clean.csv", index=False)
data_falcon9[["FlightNumber", "PayloadMass", "LaunchSite"]].head()
```
